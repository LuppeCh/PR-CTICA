```python
# Ir probando cant de features y cuando un accuracy no aumente ni subiendo ni bajando dejarlo

top16 = importancias.sort_values(ascending=False).head(16).index

print("Las 16 variables más importantes son:")
print(list(top16))

# Crear nuevos datasets solo con esas variables
X_train_top16 = X_train[top16]
X_test_top16 = X_test[top16]
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
from scipy.stats import randint

# Rango de hiperparámetros centrado en los valores por defecto
param_dist = {
    'n_estimators': randint(100, 120),
    'max_depth': randint(14, 20),
    'min_samples_split': randint(2, 4),
    'min_samples_leaf': randint(1, 3),
    'max_features': ['sqrt']
}

# Modelo base con class_weight='balanced' para mejorar recall en clase 1
modelo_base = RandomForestClassifier(
    random_state=42,
    class_weight='balanced',
    criterion='gini',
    n_jobs=-1
)

# Configuración del RandomizedSearch
random_search = RandomizedSearchCV(
    estimator=modelo_base,
    param_distributions=param_dist,
    n_iter=16,
    cv=2,
    scoring='accuracy',
    n_jobs=-1,
    random_state=42,
    verbose=1
)

# Entrenar el modelo
random_search.fit(X_train_top16, y_train)

# Mejor modelo encontrado
best_model = random_search.best_estimator_

# Predicciones
y_pred = best_model.predict(X_test_top16)

# Métricas generales
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='macro')
recall = recall_score(y_test, y_pred, average='macro')
f1 = f1_score(y_test, y_pred, average='macro')

print("\n🔍 Mejores parámetros encontrados:")
print(random_search.best_params_)
print(f"\n✅ Precisión en test con features top16: {accuracy:.4f}")
print(f"📌 Precision (macro): {precision:.4f}")
print(f"📌 Recall (macro):    {recall:.4f}")
print(f"📌 F1 Score (macro):  {f1:.4f}")

# Reporte por clase
print("\n📊 Reporte detallado por clase:")
print(classification_report(y_test, y_pred))

#los valores no superan al default
```

```python
#Uso bayesian
import optuna
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score, f1_score

def objective(trial):
    # Espacio de búsqueda alrededor de los defaults
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 90, 120),         # default = 100
        'max_depth': trial.suggest_int('max_depth', 16, 30),                # default = None, pero acotamos
        'min_samples_split': trial.suggest_int('min_samples_split', 2, 4),  # default = 2
        'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 2),    # default = 1
        'max_features': trial.suggest_categorical('max_features', ['sqrt']), # default
        'class_weight': trial.suggest_categorical('class_weight', [None, 'balanced']) # probar ambos
    }

    model = RandomForestClassifier(**params, random_state=42, n_jobs=-1)

    # Validación cruzada rápida
    score = cross_val_score(model, X_train_top16, y_train, cv=2, scoring='accuracy', n_jobs=-1).mean()
    return score

# Ejecutar la optimización
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=10, timeout=90)  # límite de 10 pruebas o 90 segundos

# Mejor modelo
best_params = study.best_params
print("🔍 Mejores parámetros encontrados:")
print(best_params)

# Entrenar modelo final con mejores parámetros
final_model = RandomForestClassifier(**best_params, random_state=42, n_jobs=-1)
final_model.fit(X_train_top16, y_train)

# Predicciones
y_pred = final_model.predict(X_test_top16)

# Métricas completas
print("\n📊 Evaluación en test:")
print(f"Accuracy:  {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred, average='macro'):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred, average='macro'):.4f}")
print(f"F1 Score:  {f1_score(y_test, y_pred, average='macro'):.4f}")

print("\n📋 Reporte por clase:")
print(classification_report(y_test, y_pred))
```

```python
def plot_confusion_matrix(y_true, y_pred, model_name):
    """Función para graficar la matriz de confusión."""
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['No comestible', 'Comestible'],
                yticklabels=['No comestible', 'Comestible'])
    plt.ylabel('Etiqueta Verdadera')
    plt.xlabel('Etiqueta Predicha')
    plt.title(f'Matriz de Confusión - {model_name}')
    plt.show()

# Hacer predicciones (si no las hiciste en el paso anterior)
logistic_predictions = logistic_model.predict(X_test)
decision_tree_predictions = decision_tree_model.predict(X_test)

# Calcular y graficar la matriz de confusión para cada modelo
plot_confusion_matrix(y_test, logistic_predictions, 'Regresión Logística')
plot_confusion_matrix(y_test, y_pred, 'Bosque Aleatorio')
plot_confusion_matrix(y_test, decision_tree_predictions, 'Árbol de Decisión')
```
